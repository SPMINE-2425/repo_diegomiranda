```python
def total_goles_equipo(df, nombre_equipo):
    """
    Calcula el total de goles de un equipo
    """
    nombre_equipo_lower = nombre_equipo.lower()
    df_equipo = df[df['Team'].str.lower() == nombre_equipo_lower]
    if df_equipo.empty:
        return f"No hay resultados para el equipo '{nombre_equipo}'."
    total_goles = df_equipo['Goals'].sum()
    return f"El total de goles del equipo {nombre_equipo} es: {total_goles}"
```

```python
def goleadores(opcion):
    """
    Devuelve los goleadores de la liga.
    """
    if opcion == 'top_10':
        gol_10= df['Goals'].iloc[10]
        return df[df['Goals'] >= gol_10]
    elif opcion == 'top_5':
        gol_5= df['Goals'].iloc[5]
        return df[df['Goals'] >= gol_5]
    elif opcion == 'top_3':
        gol_3= df['Goals'].iloc[3]
        return df[df['Goals'] >= gol_3]
    elif opcion == 'top_1':
        gol_1= df['Goals'].iloc[1]
        return df[df['Goals'] >= gol_1]
    else:
        return "Opción no válida. Por favor, elige entre 'top_10', 'top_5', 'top_3' o 'top_1'."
 
```
